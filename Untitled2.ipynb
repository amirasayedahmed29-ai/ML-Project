```python
Data = Data.drop(
    columns=[
        'RowNumber',
        'CustomerId',
        'Surname'
    ]
)

Data.head()
```

```python
Data = pd.get_dummies(
    Data,
    columns=[
        'Geography',
        'Gender'
    ],
    drop_first=True
)

Data.head()
```

```python
Data.info()

print(Data.isnull().sum())
```

```python
Predictors = Data.drop(
    columns=['Exited']
)

Response = Data[['Exited']]

print(Predictors.shape)
print(Response.shape)
```

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

PredictorsScaled = pd.DataFrame(
    scaler.fit_transform(Predictors),
    columns=Predictors.columns
)

PredictorsScaled.head()
```

```python
from sklearn.model_selection import train_test_split

Pred_train, Pred_test, Resp_train, Resp_test = train_test_split(
    PredictorsScaled,
    Response,
    test_size=0.2,
    random_state=1,
    stratify=Response
)

print(Pred_train.shape)
print(Pred_test.shape)
print(Resp_train.shape)
print(Resp_test.shape)
```

```python
from keras.models import Sequential
from keras.layers import Dense, Input
```

```python
model = Sequential()

model.add(Input(shape=(Pred_train.shape[1],)))

model.add(Dense(20, activation='relu'))

model.add(Dense(10, activation='relu'))

model.add(Dense(10, activation='relu'))

model.add(Dense(1, activation='sigmoid'))

model.summary()
```

```python
model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)
```

```python
history = model.fit(
    Pred_train,
    Resp_train,
    epochs=100,
    verbose=1,
    validation_split=0.2,
    batch_size=128
)
```

```python
y_prob = model.predict(Pred_test)

y_pred = (y_prob >= 0.5).astype(int)
```

```python
from sklearn.metrics import accuracy_score

accuracy = accuracy_score(
    Resp_test,
    y_pred
)

print(accuracy)
```

```python
from sklearn.metrics import classification_report

print(
    classification_report(
        Resp_test,
        y_pred
    )
)
```

```python
from sklearn.metrics import confusion_matrix

cm = confusion_matrix(
    Resp_test,
    y_pred
)

print(cm)
```

```python
plt.figure(figsize=(6, 5))

sns.heatmap(
    cm,
    annot=True,
    fmt='d',
    cmap='Blues'
)

plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.title('Confusion Matrix')

plt.show()
```

```python
from sklearn.metrics import roc_auc_score

roc_auc = roc_auc_score(
    Resp_test,
    y_prob
)

print(roc_auc)
```

```python
continuous_cols = [
    'CreditScore',
    'Age',
    'Tenure',
    'Balance',
    'NumOfProducts',
    'EstimatedSalary'
]

Q1 = Data[continuous_cols].quantile(0.25)

Q3 = Data[continuous_cols].quantile(0.75)

IQR = Q3 - Q1

print(IQR)
```

```python
lower_bound = Q1 - 1.5 * IQR

upper_bound = Q3 + 1.5 * IQR

print(lower_bound)
print(upper_bound)
```

```python
outlier_mask = (
    (Data[continuous_cols] < lower_bound) |
    (Data[continuous_cols] > upper_bound)
).any(axis=1)

Data_NoOutliers = Data.loc[
    ~outlier_mask
].copy()

print(Data.shape)
print(Data_NoOutliers.shape)
```

```python
Predictors2 = Data_NoOutliers.drop(
    columns=['Exited']
)

Response2 = Data_NoOutliers[
    ['Exited']
]

print(Predictors2.shape)
print(Response2.shape)
```

```python
scaler2 = MinMaxScaler()

Predictors2Scaled = pd.DataFrame(
    scaler2.fit_transform(Predictors2),
    columns=Predictors2.columns
)

Predictors2Scaled.head()
```

```python
Pred_train2, Pred_test2, Resp_train2, Resp_test2 = train_test_split(
    Predictors2Scaled,
    Response2,
    test_size=0.2,
    random_state=1,
    stratify=Response2
)

print(Pred_train2.shape)
print(Pred_test2.shape)
print(Resp_train2.shape)
print(Resp_test2.shape)
```

```python
model2 = Sequential()

model2.add(
    Input(shape=(Pred_train2.shape[1],))
)

model2.add(
    Dense(20, activation='relu')
)

model2.add(
    Dense(10, activation='relu')
)

model2.add(
    Dense(10, activation='relu')
)

model2.add(
    Dense(1, activation='sigmoid')
)

model2.summary()
```

```python
model2.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)
```

```python
history2 = model2.fit(
    Pred_train2,
    Resp_train2,
    epochs=100,
    verbose=1,
    validation_split=0.2,
    batch_size=128
)
```

```python
y_prob2 = model2.predict(
    Pred_test2
)

y_pred2 = (
    y_prob2 >= 0.5
).astype(int)
```

```python
accuracy2 = accuracy_score(
    Resp_test2,
    y_pred2
)

print(accuracy2)
```

```python
print(
    classification_report(
        Resp_test2,
        y_pred2
    )
)
```

```python
cm2 = confusion_matrix(
    Resp_test2,
    y_pred2
)

print(cm2)
```

```python
plt.figure(figsize=(6, 5))

sns.heatmap(
    cm2,
    annot=True,
    fmt='d',
    cmap='Blues'
)

plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.title('Confusion Matrix After Outlier Treatment')

plt.show()
```

```python
roc_auc2 = roc_auc_score(
    Resp_test2,
    y_prob2
)

print(roc_auc2)
```
